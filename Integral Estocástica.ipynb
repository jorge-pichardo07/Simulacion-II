```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
from scipy import stats
```

Después de importar las librerías, uso el código de movimiento browniano

```python
def brownie(T,L):
    avr = []
    w_0 = 0
    dt = T/L
    for _ in range(1,L):
        dw = np.random.normal(0,1)
        w_1 = w_0 + np.sqrt(dt)*dw
        avr.append(w_1)
        
        w_0 = w_1
    avr.insert(0,0)
    return avr
```

Y empiezo  haciendo la función para una integral de término cuadrático.
$$
\int_0^t B_{s}^2 \ ds = \sum_0^n B_{t}^2 \ (t_{i+1}-t_{i})
$$

Ahora se definen para calcular las integrales
$$
\int_0^te^{2s}ds  \sim N(0, \frac{1}{2} \cdot (e^{2t} - 1)) \quad \quad \quad \quad (1)
$$

```python
def f(x):
    return np.exp(x)
```

$$
\int_0^ts^2ds  \sim N(0, \frac{t^3}{3} ) \quad \quad \quad \quad (2)
$$

```python
def g(x):
    return x
```

$$
\int_0^t1^2ds  \sim N(0, t)) \quad \quad \quad \quad (3)
$$

```python
def v(x):
    return np.array([1 for i in range(len(x))])
```

Función para hacer la integral cuadrática:

```python
def integral(a,b,A,n):
    veo = 0
    t = np.linspace(a,b,n)
    #Si hacemos un browni
    B = np.array(brownie(1,n))
    B = A(t)**2
    for i in range(n-1):
        veo += B[i]*(t[i+1]-t[i])
    return float(veo)
```

Verifico el valor que nos dá la integral:

Para $(1)$ evaluando en $t = 1$:

```python
integral(0,1,f,1000)
```

Para $(2)$ evaluando en $t = 1$:

```python
integral(0,1,g,1000)
```

Para $(3)$ evaluando en $t = 1$:

```python
integral(0,1,v,1000)
```

Para el ejemplo del movimiento browniano, generamos una función que siga un movimiento browniano

```python
def integral_Browni(a,b,n):
    veo = 0
    t = np.linspace(a,b,n)
    #Si hacemos un browni
    B = np.array(brownie(1,n))
    B = B**2
    for i in range(n-1):
        veo += B[i]*(t[i+1]-t[i])
    return float(veo)
```

Hago una lista con todos los valores posibles en 100000 iteraciones y luego saco la media y la varianza:

```python
muestra = []
for i in range(10000):
    muestra.append(integral_Browni(0,1,10000))
media = np.mean(muestra)
varianza = np.var(muestra)    
```

Los resultados de manera analítica para $t= 1$ eran :
$$
E[\int_0^t W_sds ] = \frac{1}{2} \quad var[\int_0^t W_sds ] = \frac{1}{3}
$$
