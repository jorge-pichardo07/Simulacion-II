```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
from scipy import stats
```

```python
def brownie(T,L):
    avr = []
    w_0 = 0
    dt = T/L
    for _ in range(1,L):
        dw = np.random.normal(0,1)
        w_1 = w_0 + np.sqrt(dt)*dw
        avr.append(w_1)
        
        w_0 = w_1
    avr.insert(0,0)
    return avr
```

```python
def integral(a,b,n):
    veo = 0
    t = np.linspace(a,b,n)
    B = np.array(brownie(1,n))
    B = B**2
    for i in range(n-1):
        veo += B[i]*(t[i+1]-t[i])
    return veo
```

```python
integral(0,1,1000)
```

```python
muestra = []
for i in range(1000):
    muestra.append(integral(0,1,1000))
media = np.mean(muestra)
varianza = np.var(muestra)    
```

```python
print(f"la media es: {media}")
print(f"la varianza es: {varianza}")
```
