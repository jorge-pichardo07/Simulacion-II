Sea $S = \{0, 1\}^m$ el espacio de secuencias binarias de longitud $m$. Definimos una **secuencia buena** (o admisible) como aquella tupla binaria $x = (x_1, x_2, \dots, x_m) \in S$ que **no contiene unos adyacentes**, es decir:

$$x_i \cdot x_{i+1} = 0, \quad \forall i \in \{1, 2, \dots, m-1\}$$

El número total de secuencias buenas de longitud $m$ se denota por $N(m)$ y responde a la **Sucesión de Fibonacci**:

$$N(m) = F_{m+2}$$

## Probabilidad
Para una caminata aleatoria de longitud $N$ representada por la secuencia de variables aleatorias independientes e idénticamente distribuidas (i.i.d.) $X_1, X_2, \dots, X_N \sim \text{Bernoulli}(p=0.5)$, consideramos el espacio muestral de todas las sub-tuplas de longitud $m=4$ extraídas mediante una **ventana deslizante**.

El número total de sub-tuplas de $m=4$ bits analizadas en una cadena de longitud $N$ es:

$$T_{totales} = N - m + 1 = N - 3$$

Dado que existen $2^m = 2^4 = 16$ combinaciones posibles en total, y de las cuales exactamente $F_6 = 8$ son secuencias buenas($0000, 0001, 0010, 0100, 1000, 0101, 1010, 1001$), la probabilidad teórica de que una sub-tupla arbitraria sea buena bajo un proceso estocástico no restringido es:

$$P(\text{Tupla Buena}) = \frac{N(4)}{2^4} = \frac{8}{16} = 0.5000 \quad (50.00\%)$$

Usando la librería random, empexamos a generar la tupla buena:

```python
import random

def buena(tupla_4bits):
    for i in range(len(tupla_4bits) - 1):
        if tupla_4bits[i] == 1 and tupla_4bits[i + 1] == 1:
            return False
    return True
```

## 2. Construcción de la Caminata y Métricas Empíricas

Definimos el proceso de simulación para generar una secuencia estocástica de $N$ caracteres y evaluar mediante la función `buena()` la proporción de ventanas contiguas de tamaño $m=4$ que satisfacen la condición de no contigüidad de unos[cite: 6].

```python
def probabilidad_buenas(N_caracteres):
    # 1. Generar la caminata aleatoria de N caracteres
    cadena = [random.choice([0, 1]) for _ in range(N_caracteres)]

    # 2. Extraer todas las tuplas de tamaño 4
    total_tuplas_evaluadas = N_caracteres - 3
    tuplas_buenas = 0

    for i in range(total_tuplas_evaluadas):
        sub_tupla = tuple(cadena[i : i + 4])
        if buena(sub_tupla):
            tuplas_buenas += 1

    # 3. Calcular la probabilidad
    probabilidad = tuplas_buenas / total_tuplas_evaluadas
    porcentaje = probabilidad * 100

    return tuplas_buenas, total_tuplas_evaluadas, probabilidad, porcentaje


```

Se ejecuta la simulación con una longitud de trayectoria de $N = 10,000$ pasos para verificar la convergencia de la frecuencia relativa hacia la probabilidad teórica $P(\text{Buena}) = 0.50$.

```python

N = 10000  # Longitud N de la caminata
buenas, totales, prob, pct = probabilidad_buenas(N)


print(f"Total de tuplas analizadas: {totales}")
print(f"Tuplas 'buenas' encontradas: {buenas}")
print(f"Probabilidad de que sea buena: {prob:.4f}")
print(f"Porcentaje de tuplas buenas: {pct:.2f}%")
```
